```python
import pandas as pd

# Quick test to make sure it works
# print(pd.__version__)

path_1 = "./data/mexico-real-estate-1.csv"
# (
#     pd.read_csv(path_1).dropna().info()
# )


def read_csv(path):
    return pd.read_csv(path).dropna()

data = read_csv(path_1)
# data.head()
# data.shape
# data.info()
# data.dtypes
# data.isnull().sum()

df1 = data.assign(
price_usd = lambda x: x['price_usd'].str.replace("$", "", regex=False).str.replace(",", "", regex=False).astype(float)
)

df1.head()
df1.shape

```

```python
path_2 = "./data/mexico-real-estate-2.csv"

data2 = read_csv(path_2)
# data2.head()
df2 = data2.assign(price_usd = lambda x: x["price_mxn"].div(19)).drop(columns=["price_mxn"]).dropna()
df2.shape
# print(f"df2 type: type:{type(df2)}")
# print(f"df2 shape: {df2.shape}")
# df2.columns.tolist()
# df1.columns.tolist()
# df2["price_usd"].dtype
```

```python
path_3 = "./data/mexico-real-estate-3.csv"
data3 = read_csv(path_3)
# data3["lat-lon"].str.split(",", expand=True)[0].astype(float).head()
```

```python
# data3["place_with_parent_names"].str.split("|", expand= True).head()
```

```python
df3 = data3.assign(
    lat = lambda x : x["lat-lon"].str.split(",", expand = True)[0].astype(float),
    lon = lambda x : x["lat-lon"].str.split(",", expand = True)[1].astype(float),
    state = lambda x : x["place_with_parent_names"].str.split("|", expand= True)[2]
).dropna().drop(columns=["place_with_parent_names", "lat-lon"])

# df3.head()
df3.shape
```

The code block below helps to compare the column set of the 3 cleaned data files. 

```python
# Check if all column sets are identical
columns_match = set(df1) == set(df2) == set(df3)

if columns_match:
    print("All DataFrames have the same columns.")
else:
    print("Column mismatch detected!")
    # Optional: See the difference
    print("df1 vs df2 diff:", set(df1.columns) ^ set(df2.columns))
    print("df2 vs df3 diff:", set(df2.columns) ^ set(df3.columns))
```

```python
df = pd.concat([df1, df2, df3], ignore_index=True)
df.head()
df.shape[0]
df.shape[1]
df.isnull().sum()
```
